## Section 2: Conditional Statements
Conditional statements in Python let your program make decisions — they allow code to run only when certain conditions are true. They’re one of the most fundamental control structures in any programming language.

There are three conditional statements: if, elif, and else

* if statements are conditions that are always checked
Syntax:

    

```
if condition:
  code to execute
```



* elif statements are conditions that are checked only if an if statement is false

Syntax:

```
elif condition:
  code to execute
```
  
* else statements are conditions that are checked onyl if an if or elif statement is false

Syntax:

```
else:  
  code to execute
```

## Section 5: Creating and working with Lists/Dictionaries
A list is an ordered, mutable (changeable) collection of items. You can store numbers, strings, or even other lists inside one list.

Example:

```[1, "hello", True, ["list", 2]]```

Lists are zero based indexing, meaning the first element of a list starts at 0. You can also use negative indices to count from the end. Lists can be changed — you can add, remove, or replace elements.

Example:

```
elements = [1, 2, 3]
elements.append(4)
print(elements)
>>> 1, 2, 3, 4
```

```
elements = [1, 2, 3]
elements.remove(1)
print(elements)
>>> 2, 3
```

```
elements = [1, 2, 3]
elements[0] = 100
print(elements)
>>> 100, 2, 3
```

A dictionary is a collection of key–value pairs. It’s unordered and mutable.
Each key must be unique and immutable (like a string or number). You can access a add a key-value pair to the dictionary, retrieve a value from the dictionary, modify, or remove a value from the dictionary.

Example:

```
student = {
    "name": "John",
    "age": 22,
    "major": "Computer Science"
}

student["graduation"] = 2026

print(student)
>>> {'name': 'John', 'age': 22, 'major': 'Computer Science', 'graduation': 2026}
```

```
student = {
    "name": "John",
    "age": 22,
    "major": "Computer Science"
}
print(student.keys())
>>> ['name', 'age', 'major']

print(student.values())
>>> ['John', 22, 'Computer Science']

print(student.items())
>>> [('name', 'John'), ('age', 22), ('major', 'Computer Science')]
```

```
student = {
    "name": "John",
    "age": 22,
    "major": "Computer Science"
}
print(student['name'])
>>> 'John'

student.pop('age')
print(student)
>>> {'name': 'John', 'major': 'Computer Science', 'graduation': 2026}

student["name"] = "Susan"
print(student)
>>> {'name': 'Susan', 'major': 'Computer Science', 'graduation': 2026}
```
