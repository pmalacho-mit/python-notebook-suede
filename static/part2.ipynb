### Question 1.1
You will be given a list called __numbers__. Use a for loop to create a new list containing only the even numbers from __numbers__ and store this new list in a variable called __even_numbers__. Find the length of __even_numbers__ and store it in a variable called __length__.

````python
# START OF INSTRUCTOR CODE
numbers = [i for i in range(1, 20)] # this is a Python concept called list comprehension
# END OF INSTRUCTOR CODE

# write code here

# answer to code
even_numbers = []
length = 0
for num in numbers:

  if num % 2 == 0:
    even_numbers.append(num)
length = len(even_numbers)

print(even_numbers)
print(length)
````

### Question 1.3
You will be given a list of dictionaries called __athlete_times_list__ that represents track athletes and all the times they ran for the 100 meter dash. Use a for loop to compute the minimum time they ran, the maximum time they ran, the average time, and median time they ran. Store these values in a list of list called __descriptive_stats_list__. For each list make sure to store the different values in the following order:

* mean time
* min time
* max time
* median time

````python
# START OF INSTRUCTOR CODE
athlete_times_list = [
    {"name": "Abby", "times": [12.3, 12.1, 12.0]},
    {"name": "Becca", "times": [11.8, 12.2, 12.0]},
    {"name": "Mary", "times": [12.5, 12.4, 12.6]}
]
descriptive_stats_list = []
# END OF INSTRUCTOR CODE

# write code here

# answer to code
for athlete_times in athlete_times_list:
  length = len(athlete_times["times"])
  avg_time = sum(athlete_times["times"]) / length
  min_time = min(athlete_times["times"])
  max_time = min(athlete_times["times"])
  median = sorted(athlete_times["times"])[length // 2]
  descriptive_stats_list.append([avg_time, min_time, max_time, median])

print(descriptive_stats_list)
````

### Question 3.1
Write a function called __find_unique_and_positive_num__ that given a list of numbers, will create a dictionary with the set of unique numbers in the list, how often the numbers appeared in the list, and whether all the numbers are greater than zero. Store the result in a dictionary called __frequency_list__.

````python
# START OF INSTRUCTOR CODE
numbers = [-1, -1, 2, 3, 2, 3, 4, 5, 6, 7, -5, -5, -3]
numbers2 = [1, 2, 2, 2, 2, 4, 5, 6, 10, 10, 10, 2, 4, 5]
# END OF INSTRUCTOR CODE

# write code here

# answer to code
def find_unique_and_positive_num(num_list):
  freq_list = {"all_positive": True}

  for num in num_list:

    if num < 0:
      freq_list["all_positive"] = False

    if num not in freq_list:
      freq_list[num] = 0

    freq_list[num]+=1

  return freq_list
freq_list = find_unique_and_positive_num(numbers)
freq_list2 = find_unique_and_positive_num(numbers2)
print(freq_list)
print(freq_list2)
````

````python
# test code here
assert find_unique_and_positive_num(numbers) == {'all_positive': False, -1: 2, 2: 2, 3: 2, 4: 1, 5: 1, 6: 1, 7: 1, -5: 2, -3: 1}
assert find_unique_and_positive_num(numbers2) == {'all_positive': True, 1: 1, 2: 5, 4: 2, 5: 2, 6: 1, 10: 3}
````

### Question 3.2
Write a function called __find_honor_roll_students__ that given a list of dictionaries with student names and list of lists of scores for each student, will first update the score of a specific student (will be given the student name, the idx of the test score to update, and the new value to replace the old score). Then return a list of the students that averaged above a B on their test scores (scores of 80 and above). The function output should be stored in list called __honor_roll_students__.

**Note:** Be careful of aliasing error! Need to make a copy of list in function.

````python
# START OF INSTRUCTOR CODE
import copy

student_list = [
  {
    "name":"John",
    "scores": [100, 80, 70]
  },
  {
    "name":"Candy",
    "scores": [85, 75, 84]
  },
  {
    "name":"Peter",
    "scores": [45, 75, 84]
  },
  {
    "name":"Susan",
    "scores": [65, 73, 82]
  },
  {
    "name":"Oliver",
    "scores": [59, 57, 89]
  },
  {
    "name":"Miranda",
    "scores": [98, 75, 83]
  },
]
# END OF INSTRUCTOR CODE

# write code here

# answer to code
def find_honor_roll_students(student_scores_list, student_name, test_score_idx, new_value):
  student_scores_list = copy.deepcopy(student_scores_list)
  top_students = []

  for idx, student_dict in enumerate(student_scores_list):
    name = student_dict["name"]

    if name == student_name:
      student_scores_list[idx]["scores"][test_score_idx] = new_value
      break

  for student_scores_dict in student_scores_list:
    name, scores = student_scores_dict["name"], student_scores_dict["scores"]
    avg_score = sum(scores) / len(scores)

    if avg_score >= 80:
      top_students.append(name)

  return top_students
````

````python
# test code here
assert [name in ['John', 'Miranda'] for name in find_honor_roll_students(student_list, "Candy", 1, 67)]
assert [name in ['Candy', 'Miranda'] for name in find_honor_roll_students(student_list, "John", 0, 47)]
assert [name in ['John', 'Candy', 'Peter', 'Miranda'] for name in find_honor_roll_students(student_list, "Peter", 0, 97) ]
````
